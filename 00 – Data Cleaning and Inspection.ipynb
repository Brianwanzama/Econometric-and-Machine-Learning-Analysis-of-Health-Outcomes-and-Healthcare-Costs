### Import the dataset

#### Objective
- To inspect, clean, and prepare the synthetic health dataset for econometric and machine learning analysis, ensuring internal consistency and analytical validity.

#### Notes
- The dataset is synthetic and used for methodological demonstration. Cleaning steps follow best practices in applied health economics.


```python
import pandas as pd

# Define file path clearly
file_path = "Health_data.csv"

# Load dataset into a DataFrame
df = pd.read_csv(file_path)

# Preview the first five rows
print(df.head())
```

#### Inspect the data

```python
df.info()
```

#### Data inspection and cleaning

```python
df.shape
```

```python
#column names
df.columns
```

#### Remove the incosistencies if any

```python
# Clean and standardize column names
df.columns = (
    df.columns
      .str.strip()                # remove leading/trailing spaces
      .str.lower()                # convert to lowercase
      .str.replace(r"\s+", "_", regex=True)  # replace spaces with underscores
)
```

```python
# Inspect the data types
df.dtypes
```

#### Check the missing values

```python
df.isna().sum()
print(df.isna().mean().sort_values(ascending=False))
```

#### Check for the duplicates

```python
df.duplicated().sum()
```

#### Remove any duplicates

```python
df[df.duplicated()]
```

#### check for the values

```python
# Generate descriptive statistics for selected columns
df[['age', 'billing_amount']].describe().transpose()
```

#### Frequency distriction for the categories

```python
# Frequency counts for key categorical variables (including NaN values)
categorical_columns = [
    "gender",
    "insurance_provider",
    "admission_type",
    "medical_condition"
]

for col in categorical_columns:
    print(f"\nValue counts for '{col}':")
    print(df[col].value_counts(dropna=False))
```

#### consistency checks

```python
# Age vs admission
df.loc[df['age'] < 0]
```

```python
#cost vs admisssion type
df.groupby('admission_type')['billing_amount'].describe()
```

```python
# Save cleaned DataFrame to CSV
# Save cleaned DataFrame to CSV in the current folder
df.to_csv("health_data_cleaned.csv", index=False)
```

## Data Cleaning Summary

- The dataset contains 55500 observations and 15 variables.
- Column names were standardised for reproducibility.
-  duplicate observations were identified and handled.
- Missing values were audited and assumed to be random for demonstration purposes.
- Implausible values (if any) were addressed.
- The cleaned dataset is suitable for econometric and machine learning analysis.
